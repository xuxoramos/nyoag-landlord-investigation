```python
import duckdb
import glob

# Connect to DuckDB
con = duckdb.connect()

# Get all CSV files in the data directory
csv_files = glob.glob('data/*.csv')

# Process each CSV file
for csv_file in csv_files:
    # Extract table name from filename (without path and extension)
    table_name = csv_file.split('/')[-1].replace('.csv', '')
    
    # Read CSV with all columns as VARCHAR
    con.execute(f"""
        CREATE TABLE {table_name} AS 
        SELECT * FROM read_csv_auto('{csv_file}', ALL_VARCHAR=TRUE)
    """)
    
    # Export to Parquet
    con.execute(f"COPY {table_name} TO 'data/{table_name}.parquet' (FORMAT PARQUET)")
    
    print(f"Exported {table_name} to data/{table_name}.parquet")

con.close()
```
